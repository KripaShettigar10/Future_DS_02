```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Telco_Customer_Churn.csv")
```

```python
df.head()
```

```python
print("Dataset shape:",df.shape)
print("Column names:")
print(df.columns.tolist())
```

```python
df.info()
```

```python
print("Missing Values in each column:")
print(df.isnull().sum())
```

```python
print("Number of duplicate rows:")
print(df.duplicated().sum())
```

```python
df. describe(include="all")
```

```python
print(df["Churn"].value_counts())

print("\nChurn percentage:")
print(df["Churn"].value_counts(normalize=True).mul(100).round(2))
```

```python
print(df.dtypes)
```

```python
print(df["TotalCharges"].head(10))
print("Data type:", df["TotalCharges"].dtype)
```

```python
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"], errors="coerce"
)

print("Missing values after conversion:")
print(df["TotalCharges"].isnull().sum())
```

```python
print(df["TotalCharges"].dtype)
print(df["TotalCharges"].isnull().sum())
```

```python
df[df["TotalCharges"].isnull()]
```

```python
df.loc[df["TotalCharges"].isnull(), ["customerID", "tenure", "MonthlyCharges", "TotalCharges", "Churn"]]
```

```python
df["TotalCharges"] = df["TotalCharges"].fillna(0)
```

```python
print("Missing values after cleaning:")
print(df.isnull().sum().sum())

print("\nTotalCharges data type:")
print(df["TotalCharges"].dtype)
```

```python
total_customers = df.shape[0]

print("Total Number of Customers:", total_customers)
```

```python
churn_counts = df["Churn"].value_counts()

print("Customer Churn Count:")
print(churn_counts)
```

```python
churn_percentage = df["Churn"].value_counts(normalize=True) * 100

print("Customer Churn Percentage:")
print(churn_percentage.round(2))
```

```python
total_customers = len(df)

churned_customers = (df["Churn"] == "Yes").sum()

churn_rate = (churned_customers / total_customers) * 100

print("Total Customers:", total_customers)
print("Customers Who Left:", churned_customers)
print("Overall Churn Rate:", round(churn_rate, 2), "%")
```

```python
sns.countplot(data=df, x="Churn")

plt.title("Customer Churn Distribution")
plt.xlabel("Churn Status")
plt.ylabel("Number of Customers")
plt.show()
```

```python
contract_churn = pd.crosstab(
    df["Contract"],
    df["Churn"]
)

print(contract_churn)
```

```python
contract_churn_rate = pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100

print(contract_churn_rate.round(2))
```

```python
contract_churn_rate["Yes"].plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Customer Churn Rate by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
tenure_churn = df.groupby("Churn")["tenure"].mean()

print("Average Customer Tenure by Churn Status:")
print(tenure_churn.round(2))
```

```python
tenure_churn.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Average Customer Tenure by Churn Status")
plt.xlabel("Churn Status")
plt.ylabel("Average Tenure (Months)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
monthly_charges = df.groupby("Churn")["MonthlyCharges"].mean()

print("Average Monthly Charges by Churn Status:")
print(monthly_charges.round(2))
```

```python
monthly_charges.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Average Monthly Charges by Churn Status")
plt.xlabel("Churn Status")
plt.ylabel("Average Monthly Charges")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
total_customers = len(df)

churned_customers = (df["Churn"] == "Yes").sum()

retained_customers = (df["Churn"] == "No").sum()

churn_rate = (churned_customers / total_customers) * 100

retention_rate = (retained_customers / total_customers) * 100

print("CUSTOMER RETENTION AND CHURN SUMMARY")
print("------------------------------------")
print("Total Customers:", total_customers)
print("Churned Customers:", churned_customers)
print("Retained Customers:", retained_customers)
print("Overall Churn Rate:", round(churn_rate, 2), "%")
print("Overall Retention Rate:", round(retention_rate, 2), "%")
```
